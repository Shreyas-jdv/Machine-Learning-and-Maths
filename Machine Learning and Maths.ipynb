## Linear Algebra 

### Why do we need vectors?
* **Traditional Approach (Hand-crafted feature learning):** Input data goes through feature extraction (such as SIFT, SURF, HOG) to generate hand-crafted feature vectors, which are then fed into learning algorithms (SVM, XGBoost, k-NN, decision trees) for prediction.

### Vectors
* An $n$-dimensional vector $x \in \mathbb{R}^n$ is written as:
  $$x = \begin{bmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{bmatrix}$$
* $x_i$ represents the $i$-th element of the vector.

## Vector Operations and Norms

### Operations on Vectors
* **Transpose:** Converts a column vector to a row vector: 
  $$x^T = \begin{bmatrix} x_1 & x_2 & \dots & x_n \end{bmatrix}$$
* **Addition:** Element-wise addition of vectors: 
  $$x + y = \begin{bmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{bmatrix} + \begin{bmatrix} y_1 \\ y_2 \\ \vdots \\ y_n \end{bmatrix} = \begin{bmatrix} x_1 + y_1 \\ x_2 + y_2 \\ \vdots \\ x_n + y_n \end{bmatrix}$$
* **Inner Product (Dot Product):** 
  $$x^T y = x_1 y_1 + x_2 y_2 + \dots + x_n y_n$$

### Vector p-Norms
* Defined as $\|x\|_p = (|x_1|^p + |x_2|^p + \dots + |x_n|^p)^{\frac{1}{p}}$ with $p > 0$.
  * **$p = 0$:** Counts the number of non-zero elements in $x$ (sparsity).
  * **$p = 1$:** Sum of absolute values of elements.
  * **$p = 2$:** Length (Euclidean norm) of the vector.
* **Properties:** $\|x\|_p \ge 0$ (and $\|x\|_p = 0$ if and only if $x = 0$), $\|kx\|_p = |k| \|x\|_p$ for any scalar $k \in \mathbb{R}$, and the triangle inequality $\|x + y\|_p \le \|x\|_p + \|y\|_p$.

> * In NumPy and PyTorch, pay close attention to vector shapes (`(n,)` vs `(n, 1)`). Misaligned dimensions are the #1 source of matrix multiplication errors!

## Distance and Cosine Similarity

### Euclidean Distance and Cosine Similarity
* **Euclidean Distance:** 
  $$\|x - y\| = \sqrt{(x_1 - y_1)^2 + \dots + (x_n - y_n)^2}$$
* **Cosine Similarity:** 
  $$\cos(\theta) = \cos(x, y) = \frac{x^T y}{\|x\| \|y\|} = \frac{x^T y}{\sqrt{x^T x}\sqrt{y^T y}}$$
  * $-1 \le \cos(x, y) \le 1$.
* **Cosine Distance:** 
  $$d_{\text{cosine}}(x, y) = 1 - \cos(x, y)$$

### Example: Word2Vec Cosine Similarity
* Assume Word2Vec transforms words *King*, *Queen*, and *Football* into vectors:
  * $v_K = [1, -2, 1]^T$
  * $v_Q = [0.9, -2.2, 1.2]^T$
  * $v_F = [1, 2.1, -0.9]^T$
* Calculating $\cos\_sim(\text{King}, \text{Queen}) \approx 0.996$ and $\cos\_sim(\text{Football}, \text{Queen}) \approx -0.72$ shows that **King is more similar to Queen than Football**.

## Matrices and Tensors

### Matrix Definition and Operations
* A matrix $A \in \mathbb{R}^{m \times n}$ has $m$ rows and $n$ columns.
* **Transpose:** $C = A^T \in \mathbb{R}^{n \times m}$ where $C_{ij} = A_{ji}$ for all valid $i, j$.
* **Scalar Multiplication & Addition:** Element-wise scaling ($aA$) and matrix addition ($A + B$).
* **Matrix Multiplication:** Given $A \in \mathbb{R}^{m \times n}$ and $B \in \mathbb{R}^{n \times p}$, the product $C = AB \in \mathbb{R}^{m \times p}$ is computed as $c_{ij} = \sum_{k=1}^n a_{ik} b_{kj} = A[i, :]B[:, j]$ (dot product of row $i$ in $A$ and column $j$ in $B$).

### Tensors
* A tensor is a multi-dimensional array:
  * **0D Tensor:** Scalar.
  * **1D Tensor:** Vector.
  * **2D Tensor:** Matrix.
  * **3D Tensor:** 3D hyperrectangle / cube of numerals.

## Information Theory

### Basic Information-Theoretic Quantities
Given two discrete distributions $p = [p_i]_{i=1}^d$ ($\sum p_i = 1$) and $q = [q_i]_{i=1}^d$ ($\sum q_i = 1$):
* **Kullback-Leibler (KL) Divergence:** 
  $$\text{KL}(p, q) = \sum_{i=1}^d p_i \log \frac{p_i}{q_i}$$ 
  (where $\text{KL}(p, q) \ge 0$, and $\text{KL}(p, q) = 0$ if and only if $p = q$).
* **Shannon Entropy:** 
  $$H(p) = -\sum_{i=1}^d p_i \log p_i$$
* **Cross-Entropy (CE) Divergence:** 
  $$\text{CE}(p, q) = -\sum_{i=1}^d p_i \log q_i = \text{KL}(p, q) + H(p)$$

## Machine Learning 



### Definitions of Machine Learning
* *"Field of study that gives computers the ability to learn without being explicitly programmed"* (Arthur Samuel, 1959).
* *"A computer program is said to learn from experience $E$ with respect to some class of tasks $T$ and performance measure $P$, if its performance at tasks in $T$, as measured by $P$, improves with experience $E$"* (Tom Mitchell, 1997).

### Elements of Machine Learning
1. **Data:** Vector $x \in \mathbb{R}^d$ (image, sentence, speech) and ground-truth label $y \in Y$.
2. **Model:** Mapping $f: X \to Y$ (Classification: $Y = \{1, \dots, M\}$, Regression: $Y = \mathbb{R}$).
3. **Assessment:** Performance metrics such as Accuracy, Precision, Recall, and F-score.
   $$\text{Accuracy} = \frac{\text{# Correct Predictions}}{\text{# All Predictions}}$$

## Supervised Learning: Classification and Regression

### Classification
* **Label Set:** $Y = \{1, 2, \dots, M\}$.
* **Training Set:** $D = \{(x_1, y_1), \dots, (x_N, y_N)\}$ where $x_i \in \mathbb{R}^d$ and $y_i \in Y$.
* **Goal:** Learn a model $f: X \to Y$ to assign a discrete class label to an input data point $x$.

### Regression
* **Label Set:** $Y = \mathbb{R}$ (continuous values).
* **Training Set:** $D = \{(x_1, y_1), \dots, (x_N, y_N)\}$ where $x_i \in \mathbb{R}^d$ and $y_i \in \mathbb{R}$.
* **Goal:** Learn a model $f: X \to Y$ to predict a continuous real-valued output $\hat{y} \in \mathbb{R}$.

## Discriminative Machine Learning and Softmax

### Discriminative Models
* A discriminative ML model maps a data point $x$ to discriminative scores/logits $h_1(x), \dots, h_M(x) \in \mathbb{R}$.
* **Prediction:** $\hat{y} = \arg\max_{1 \le m \le M} h_m(x)$.

### Softmax Activation Function
* Transforms real-valued discriminative scores into discrete probabilities:
  $$\text{softmax}(h)_m = \frac{\exp(h_m)}{\sum_{i=1}^M \exp(h_i)}$$
* Maps scores from $(-\infty, +\infty)$ to probability values in $[0, 1]$ while preserving order and summing to 1.

## Cross-Entropy Loss & Logistic Regression

### Cross-Entropy Loss
* Ground-truth labels can be represented as a one-hot vector $1_y$.
* **Loss Function:** 
  $$l(y, \hat{y}) = \text{CE}(1_y, p) = -\log p_y$$
* **Training Set Loss:** 
  $$L(D, h) = \frac{1}{N} \sum_{i=1}^N \text{CE}(1_{y_i}, p(x_i)) = -\frac{1}{N} \sum_{i=1}^N \log p_{y_i}(x_i)$$

### Logistic Regression (Simple Feed-Forward Neural Network)
* $x \in \mathbb{R}^d$, weights $W \in \mathbb{R}^{d \times 2}$, bias $b \in \mathbb{R}^{1 \times 2}$:
  $$h(x) = xW + b$$
  $$p(x) = \text{softmax}(h(x))$$

## Logistic Regression Pipeline

* **Mini-batch Processing:** Data is batched (e.g., 64 examples per batch).
* **Forward Computation:** Compute logits $h(x)$ and prediction probabilities $p(x)$ using model parameters $\theta = [W, b]$.
* **Batch Loss:** 
  $$\text{batch\_loss}(\theta) = \frac{1}{64} \sum_{i=1}^{64} \text{CE}(1_{y_i}, p(x_i, \theta))$$
* **Optimization:** Use optimizers (Adam, SGD, RMSprop, etc.) to minimize the batch loss and update parameters $\theta$.

> **🔬 Experimentation Ideas:**
> * Try swapping out `torch.optim.SGD` for `torch.optim.Adam` in the pipeline and observe how many fewer epochs it takes for the training batch loss to converge!
